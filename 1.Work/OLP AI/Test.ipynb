```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.stack([a, b], 1)
print(a.shape, b.shape, c.shape)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6, 7])
c = np.concatenate([a, b], axis=0)
print(a.shape)
print(b.shape)
print(c.shape)
```
